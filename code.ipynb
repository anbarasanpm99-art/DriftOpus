```python
import pandas as pd
import numpy as np
```

```python
class DrifyOpus:
    def __init__(self, source: pd.DataFrame, streams: pd.DataFrame, target_column: str): # Outlet_Type
        self.source = source
        self.streams = streams
        self.target_column = target_column

    """"Data Frame Check : To check the type of the the input dataframes"""

    def is_dataframe(self):

        if isinstance(self.source, pd.DataFrame) and isinstance(self.streams, pd.DataFrame):
            return True
        else:
            return 
        
        

    def shap(self):
        return self.source.shape, self.streams.shape,self.target_column
```

```python
source=pd.read_csv("data\source.csv")
stream=pd.read_csv("data\stream.csv")
target_column="Outlet_Type"

dft_ops=DrifyOpus(source,stream,target_column)
dft_ops.is_dataframe()
```
